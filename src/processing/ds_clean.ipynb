## FUNCIÓN PARA DESCODIFICAR

```python
def decodif(df_original: pd.DataFrame, df_descod: list[pd.DataFrame]) -> pd.DataFrame:
    """
    Descodifica una o varias columnas de un DataFrame usando tablas descodificadoras de referencia.

    Parameters: 
        df_original (pd.DataFrame): DataFrame original. 
        df:descod (list[pd.DataFrame]): Lista de configuraciones de descodificación.
 
    Returns: 
        pd.DataFrame: DataFrame original con la/s columna/s descodificada/s.
    
    """
    
    df_copy = df_original.copy()
    col_order = df_copy.columns
    
    len_df_descod = len(df_descod)



    if len_df_descod == 1:
        if df_descod[0].columns[0] in col_order:
            df_copy = df_copy.merge(df_descod[0], on = df_descod[0].columns[0], how = 'left').drop(df_descod[0].columns[0], axis = 1)
            df_copy = df_copy.rename(columns={df_descod[0].columns[1]:df_descod[0].columns[0]})[col_order]
        else:
            print(f'No se encuentra {df_descod[0].columns[0]} entre las variables del DataFrame. No se puede descodificar.')
    elif len_df_descod > 1: 
        for descod in df_descod:
            if  descod.columns[0] in col_order:
                df_copy = df_copy.merge(descod, on = descod.columns[0], how = 'left').drop(descod.columns[0], axis = 1)
                df_copy = df_copy.rename(columns = {descod.columns[1]:descod.columns[0]})[col_order]
            else: 
                print(f'No se encuentra {descod.columns[0]} entre las variables del DataFrame. No se puede descodificar.')
    else: 
        print("ERROR: no hay dataframe de descodificación o") 
    
    return df_copy
```

```python
def cambio_nombre_col(df: pd.DataFrame, dict_cols: dict) -> pd.DataFrame:
    
    df_copy = df.copy()
    col = df_copy.columns

    cols_no_encontradas = [x for x in dict_cols.keys() if x not in col]
    if cols_no_encontradas:
        print(f"Columnas no encontradas del diccionario: {cols_no_encontradas}")
    else: 
        print('Todas las claves del diccionario se encuentran como columnas en el dataframe.')
    
    df_copy.rename(dict_cols, axis=1, inplace=True)
    
    return df_copy
```
